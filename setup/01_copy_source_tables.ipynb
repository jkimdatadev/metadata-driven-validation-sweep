# Copy source tables
Copies selected TPC-H SF1 tables into a writable schema for validation development.


```python
%sql
-- Create the project catalog and development schema
create catalog if not exists validation_sweep;
create schema if not exists validation_sweep.dev;
```

```python
%sql
-- Copy TPC-H SF1 source tables into the development schema
-- Starting with one single-column key (customer) and one composite key (partsupp)
create table if not exists validation_sweep.dev.customer as
select * from samples.tpch_sf1.customer;

create table if not exists validation_sweep.dev.partsupp as
select * from samples.tpch_sf1.partsupp;
```

```python
# Verify copies match the source in row count and schema
for t in ["customer", "partsupp"]:
    src = spark.table(f"samples.tpch_sf1.{t}")
    dst = spark.table(f"validation_sweep.dev.{t}")
    print(f"{t:10} rows match: {src.count() == dst.count()}   "
          f"schema match: {src.schema == dst.schema}")
```
