# Test error handling
Temporarily registers an invalid key column to confirm that the validation runner reports an `ERROR` row and continues with the remaining tables.

Run the cells in order. The last cell restores the registry.

```python
%sql
-- Add a temporary registry entry with a key column that doesn't exist
insert into validation_sweep.metadata.pk_registry values
  ('validation_sweep', 'dev', 'customer', array('c_invalidkey'), true, current_timestamp());
```

```python
%sql
-- Remove the temporary entry
delete from validation_sweep.metadata.pk_registry
where schema_name = 'dev'
  and table_name = 'customer'
  and array_contains(key_columns, 'c_invalidkey');
```
